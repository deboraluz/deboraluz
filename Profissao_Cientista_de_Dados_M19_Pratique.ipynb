**Efetividade de Duas Estratégias de Ensino**

Imagine que uma escola esteja avaliando a eficácia de duas estratégias de ensino de matemática para alunos do ensino médio. Eles querem determinar se há uma diferença significativa no desempenho médio dos alunos entre as duas estratégias.

# **Hipóteses:**

* Hipótese nula (H0): A média das notas dos alunos na estratégia A é igual à média das notas dos alunos na estratégia B.
* Hipótese alternativa (H1): A média das notas na Estratégia B é maior do que a média das notas na Estratégia A.

# **Dados:**

* Amostra da Estratégia A: Notas de 50 alunos que receberam a Estratégia A.
* Amostra da Estratégia B: Notas de 50 alunos que receberam a Estratégia B.

Usaremos um teste Z para comparar as médias das notas entre as duas amostras.

Se o p-valor do teste Z for menor que um nível de significância pré-determinado (por exemplo, α = 0.05), rejeitamos a hipótese nula e concluímos que há uma diferença significativa nas médias das notas entre as duas estratégias de ensino.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

Os dados são criados a seguir:

```python
# Definindo médias e desvios padrão para as notas nas duas estratégias
media_estrategia_A = 70
desvio_padrao_estrategia_A = 10

media_estrategia_B = 75
desvio_padrao_estrategia_B = 12

# Gerando as amostras de notas para cada estratégia de ensino da nossa base
np.random.seed(0)  # Para garantir a reprodutibilidade dos resultados
amostra_estrategia_A = np.random.normal(loc=media_estrategia_A, scale=desvio_padrao_estrategia_A, size=50)
amostra_estrategia_B = np.random.normal(loc=media_estrategia_B, scale=desvio_padrao_estrategia_B, size=50)

print("Notas da Estratégia A:", amostra_estrategia_A[:5])
print("Notas da Estratégia B:", amostra_estrategia_B[:5])
```

# 1) De acordo com as informações analisadas o nosso teste é unilateral á direita, esquerda ou bicaudal? Justifique.




```python
# Sua justificativa aqui
```

O teste é **unilateral à direita**, porque a hipótese alternativa afirma especificamente que a média da Estratégia B é **maior** que a média da Estratégia A (H1: μB > μA). Portanto, estamos interessados apenas em verificar se a diferença ocorre no sentido B > A. O nível de significância adotado é α = 0,05.

# 2) Calcule as médias para as duas amostragens e as variâncias. Quais insights você pode retirar comparando os dados?

```python
#Seu código aqui
media_A = np.mean(amostra_estrategia_A)
media_B = np.mean(amostra_estrategia_B)
variancia_A = np.var(amostra_estrategia_A, ddof=1)
variancia_B = np.var(amostra_estrategia_B, ddof=1)

print(f"Média da Estratégia A: {media_A:.2f}")
print(f"Variância da Estratégia A: {variancia_A:.2f}")
print(f"Média da Estratégia B: {media_B:.2f}")
print(f"Variância da Estratégia B: {variancia_B:.2f}")

print("\nInsight:")
print("A Estratégia B apresentou média maior que a Estratégia A, indicando um desempenho médio superior na amostra.")
print("A variância da Estratégia A foi maior que a da Estratégia B, indicando maior dispersão das notas na Estratégia A nesta amostra.")
```

A média da Estratégia B é maior (aproximadamente 74,75) do que a da Estratégia A (aproximadamente 71,41), uma diferença de cerca de 3,34 pontos. A variância da Estratégia A é aproximadamente 129,27, enquanto a da Estratégia B é aproximadamente 110,47. Assim, além de apresentar uma média maior, a Estratégia B teve menor dispersão das notas nessa amostra. A diferença de médias, porém, precisa ser avaliada pelo teste de hipótese para verificar se é estatisticamente significativa.

# 3) Imprima os resultados da estatística do teste Z, p value e indique se rejeitaremos ou não a hipótese nula. Justifique sua resposta.

```python
#Seu código aqui
n_A = len(amostra_estrategia_A)
n_B = len(amostra_estrategia_B)

# Estatística Z para a diferença entre duas médias independentes
erro_padrao = np.sqrt(variancia_A / n_A + variancia_B / n_B)
estatistica_Z = (media_B - media_A) / erro_padrao

# Teste unilateral à direita: H1: μB > μA
p_value = stats.norm.sf(estatistica_Z)
alpha = 0.05

print(f"Estatística do teste Z: {estatistica_Z:.4f}")
print(f"p-value: {p_value:.4f}")

if p_value < alpha:
    print("Decisão: rejeitamos H0.")
    print("Justificativa: o p-value é menor que 0,05, indicando evidência estatística suficiente para concluir que a média da Estratégia B é maior que a da Estratégia A.")
else:
    print("Decisão: não rejeitamos H0.")
    print("Justificativa: o p-value é maior que 0,05. Portanto, não há evidência estatística suficiente, ao nível de 5%, para concluir que a média da Estratégia B é maior que a da Estratégia A.")
```

O resultado esperado é Z ≈ 1,5267 e p-value ≈ 0,0634. Como o p-value é maior que α = 0,05, não rejeitamos a hipótese nula. Apesar de a média da Estratégia B ser maior na amostra, a diferença não é estatisticamente significativa ao nível de 5%. Portanto, com esses dados, não podemos afirmar que a Estratégia B produz uma média de notas maior que a Estratégia A.

# 4) Para finalizar monte o gráfico da distribuição da estatística do nosso teste Z e explique o que pode ser observado através dele.

```python
#Seu código aqui
# Valor crítico para um teste unilateral à direita com alpha = 0,05
z_critico = stats.norm.ppf(1 - alpha)

x = np.linspace(-4, 4, 1000)
y = stats.norm.pdf(x)

plt.figure(figsize=(10, 5))
plt.plot(x, y, label='Distribuição Normal Padrão')

# Área crítica
x_crit = np.linspace(z_critico, 4, 300)
plt.fill_between(x_crit, stats.norm.pdf(x_crit), alpha=0.35, label='Área crítica (α = 0,05)')

# Área do p-value
x_p = np.linspace(estatistica_Z, 4, 300)
plt.fill_between(x_p, stats.norm.pdf(x_p), alpha=0.35, label='p-value')

plt.axvline(z_critico, linestyle='--', label=f'Z crítico = {z_critico:.3f}')
plt.axvline(estatistica_Z, linestyle=':', label=f'Z observado = {estatistica_Z:.3f}')

plt.title('Teste Z unilateral à direita')
plt.xlabel('Estatística Z')
plt.ylabel('Densidade')
plt.legend()
plt.grid(alpha=0.2)
plt.show()

print(f"Z crítico: {z_critico:.4f}")
print(f"Z observado: {estatistica_Z:.4f}")
print(f"p-value: {p_value:.4f}")
```

A linha do Z observado fica à esquerda do valor crítico de aproximadamente 1,645. A área crítica está na cauda direita da distribuição, enquanto o p-value corresponde à área à direita do Z observado. Como o Z observado não alcança a região crítica e o p-value (≈ 0,0634) é maior que 0,05, o gráfico confirma visualmente a decisão de não rejeitar H0.
