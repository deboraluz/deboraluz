# 1 - Traga a média e a mediana dos preços - coluna Preco_Normal - por categoria de produto.
# Identifique as categorias que parecem ter um valor de média abaixo ou acima da mediana.

```python
#Media
df.groupby('Categoria')['Preco_Normal'].mean().reset_index().sort_values(by='Preco_Normal', ascending=False)
```

```python
#Mediana
df.groupby('Categoria')['Preco_Normal'].median().reset_index().sort_values(by='Preco_Normal', ascending=False)
```

Apenas a categoria "comidas-preparadas" possui a média abaixo da mediana, as demais todas a média está acima da mediana.

Isso sugere que existem produtos de 'Lácteos' com preços consideravelmente mais altos que puxam a média para cima, enquanto a maioria dos produtos está concentrada em preços mais baixos.

2.  **Beleza e Cuidados Pessoais (belleza-y-cuidado-personal)**:

3.  **Congelados (congelados)**:
    *   **Desvio Padrão:** 2111.54 (o terceiro maior)
    *   **Média:** 2108.04
    *   **Mediana:** 1519.00

# 3 - Plot um boxplot da distribuição do Preco_Normal para a categoria que você identificou que tem o maior desvio padrão. Como é a distribuição desses dados segundo o boxplot? Você identifica muitos outliers?

Dica: Para trazer apenas os dados da categoria que você deseja você pode usar o df.loc[df['Categoria'] == 'CATEGORIA ESCOLHIDA'

```python
#BoxPlot
lacteos = df.loc[df['Categoria'] == 'lacteos']

fig = px.box(lacteos,
             y='Preco_Normal',
             color_discrete_sequence=['royalblue'],
             title='Boxplot do Preço Normal - Categoria Lácteos')

fig.update_layout(yaxis_title='Preço Normal',
                  xaxis_title='')
fig.show()
```

O boxplot da categoria lacteos mostra que a distribuição dos preços é assimétrica à direita. A mediana é de aproximadamente 989, enquanto o Q3 é 2619, e o limite superior é cerca de 5934. Acima desse valor há diversos outliers, indicando a presença de muitos produtos com preços significativamente maiores que a maior parte da categoria.

Esses outliers elevam a média (2385,22) em relação à mediana (989,00) e explicam o fato de a categoria apresentar o maior desvio padrão (3925,82) entre todas as categorias analisadas. Portanto, conclui-se que os preços dos produtos da categoria lacteos possuem alta variabilidade e são fortemente influenciados por produtos de alto valor.

# 4 - Plote um gráfico de barras onde temos a média de descontos por categoria.

```python
#Grafico de Barras
media_desconto = (
    df.groupby('Categoria')['Desconto']
      .mean()
      .reset_index()
      .sort_values(by='Desconto', ascending=False))

fig = px.bar(
    media_desconto,
    x='Categoria',
    y='Desconto',
    title='Média de Descontos por Categoria',
    text_auto='.2f')

fig.update_layout(
    xaxis_title='Categoria',
    yaxis_title='Média de Desconto')

fig.show()
```

# 5 - Plote um gráfico de mapa interativo agrupando os dados por categoria, marca e trazendo a média de desconto.

```python
#Treemap
media_desconto = (df.groupby(['Categoria', 'Marca'])
      .agg(
          Media_Desconto=('Desconto', 'mean'),
          Quantidade=('Desconto', 'count'))
      .reset_index())

fig = px.treemap(media_desconto,
    path=['Categoria', 'Marca'],
    values='Quantidade',
    color='Media_Desconto',
    color_continuous_scale='Blues',
    title='Média de Desconto por Categoria e Marca')

fig.show()
```
